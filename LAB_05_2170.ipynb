```python
import gymnasium as gym
import random


def random_agent(environment_name, episodes=10):

    # Step 1: Create environment
    env = gym.make(environment_name)

    rewards = []

    print("\n====================================")
    print("Environment:", environment_name)
    print("====================================")

    # Run multiple episodes
    for episode in range(1, episodes + 1):

        # Step 2: Reset environment
        state, info = env.reset()

        total_reward = 0
        terminated = False
        truncated = False

        # Continue until episode ends
        while not (terminated or truncated):

            # Step 4: Identify available actions
            number_of_actions = env.action_space.n

            # Step 5: Select random action
            action = random.randrange(number_of_actions)

            # Step 6: Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # Step 7: Add reward
            total_reward += reward

            # Update state
            state = next_state

        # Store episode reward
        rewards.append(total_reward)

        print(
            "Episode", episode,
            ": Total Reward =", total_reward
        )

    # Step 9: Calculate average reward
    average_reward = sum(rewards) / len(rewards)

    print("\nTotal Rewards:", rewards)
    print("Average Reward:", average_reward)

    # Close environment
    env.close()


# Run Random Agent on CartPole
random_agent("CartPole-v1", 10)

# Run Random Agent on MountainCar
random_agent("MountainCar-v0", 10)

# Run Random Agent on FrozenLake
random_agent("FrozenLake-v1", 10)
```
