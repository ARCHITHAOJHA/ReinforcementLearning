Assignment1: Calculation of State and Action Values Using Bellman Expectation Equations for
stochastic policy and deterministic environment

```python

# Assignment 1: Stochastic Policy, Deterministic Environment

gamma = 0.9

# Bellman equations:
# V0 = -1 + 0.81 * V1
# V1 = 89.9 + 0.09 * V0

# Solve algebraically:
# V0 = -1 + 0.81 * (89.9 + 0.09 * V0)
# 0.9271 * V0 = 71.819

V0 = 71.819 / 0.9271
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q0_down = -1 + gamma * V1
Q0_right = -1

Q1_right = 100
Q1_up = -1 + gamma * V0

# Verify Bellman expectation equations
V0_check = 0.9 * Q0_down + 0.1 * Q0_right
V1_check = 0.9 * Q1_right + 0.1 * Q1_up

print("ASSIGNMENT 1 RESULTS")
print(f"V(S0) = {V0:.2f}")
print(f"V(S1) = {V1:.2f}")

print(f"Q(S0, Down) = {Q0_down:.2f}")
print(f"Q(S0, Right) = {Q0_right:.2f}")
print(f"Q(S1, Right) = {Q1_right:.2f}")
print(f"Q(S1, Up) = {Q1_up:.2f}")

print("\nVERIFICATION")
print(f"V(S0) from Q values = {V0_check:.2f}")
print(f"V(S1) from Q values = {V1_check:.2f}")
```

Assignment 2: Calculation of State and Action Values Using Bellman Expectation Equations for
stochastic environment and deterministic policy

```python

# Assignment 2: Deterministic Policy, Stochastic Environment

gamma = 0.9
theta = 0.000001

# Initial state values
V0 = 0.0
V1 = 0.0
iterations = 0

# Iterative policy evaluation
while True:
    # Use old values for both updates
    old_V0 = V0
    old_V1 = V1

    # S0: Down -> Wall (0.1), S1 (0.9)
    new_V0 = (
        0.1 * (-1 + gamma * 0)
        + 0.9 * (-1 + gamma * old_V1)
    )

    # S1: Right -> Goal (0.9), S0 (0.1)
    new_V1 = (
        0.9 * (100 + gamma * 0)
        + 0.1 * (-1 + gamma * old_V0)
    )

    delta = max(
        abs(new_V0 - old_V0),
        abs(new_V1 - old_V1)
    )

    V0 = new_V0
    V1 = new_V1
    iterations += 1

    if delta < theta:
        break

# Calculate action values after convergence
Q0_down = (
    0.1 * (-1 + gamma * 0)
    + 0.9 * (-1 + gamma * V1)
)

Q0_right = -1 + gamma * 0

Q1_right = (
    0.9 * (100 + gamma * 0)
    + 0.1 * (-1 + gamma * V0)
)

Q1_up = -1 + gamma * V0

# Display results
print("ASSIGNMENT 2 RESULTS")
print("Iterations:", iterations)
print(f"V(S0) = {V0:.6f}")
print(f"V(S1) = {V1:.6f}")

print(f"Q(S0, Down) = {Q0_down:.6f}")
print(f"Q(S0, Right) = {Q0_right:.6f}")
print(f"Q(S1, Right) = {Q1_right:.6f}")
print(f"Q(S1, Up) = {Q1_up:.6f}")

# Policy consistency checks
print("\nPOLICY CONSISTENCY CHECKS")
print(f"V(S0) == Q(S0, Down): "
      f"{abs(V0 - Q0_down) < theta}")
print(f"V(S1) == Q(S1, Right): "
      f"{abs(V1 - Q1_right) < theta}")
```
