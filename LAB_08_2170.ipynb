```python
# Assignment 1
# Calculate State Value and Action Value using Trajectory Method
# No Bellman Equation

GAMMA = 0.9

# -----------------------------
# Grid World
# -----------------------------

states = ["S0", "S1", "S3", "S4", "G"]

actions = ["Up", "Down", "Left", "Right"]

# State positions
positions = {
    "S0": (0, 0),
    "S1": (0, 1),
    "G":  (0, 2),
    "S3": (1, 1),
    "S4": (1, 2)
}

# Obstacle position
obstacle = (1, 0)

# Policy
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}


# -----------------------------
# Find next state
# -----------------------------

def get_next_state(state, action):

    row, col = positions[state]

    if action == "Up":
        new_position = (row - 1, col)

    elif action == "Down":
        new_position = (row + 1, col)

    elif action == "Left":
        new_position = (row, col - 1)

    elif action == "Right":
        new_position = (row, col + 1)

    # Check obstacle
    if new_position == obstacle:
        return "OBSTACLE"

    # Check outside grid
    if new_position[0] < 0 or new_position[0] > 1:
        return state

    if new_position[1] < 0 or new_position[1] > 2:
        return state

    # Find state at new position
    for s, pos in positions.items():
        if pos == new_position:
            return s

    return state


# -----------------------------
# Get reward
# -----------------------------

def get_reward(next_state):

    if next_state == "G":
        return 10

    elif next_state == "OBSTACLE":
        return -5

    else:
        return -1


# ==========================================================
# STATE VALUE USING TRAJECTORY METHOD
# ==========================================================

def state_value(start_state):

    state = start_state
    total_return = 0
    discount = 1

    print("\nTrajectory starting from", start_state)

    while state != "G":

        action = policy[state]

        next_state = get_next_state(state, action)

        reward = get_reward(next_state)

        print(
            state, "--", action,
            "-->", next_state,
            " Reward =", reward
        )

        # Discounted return
        total_return += discount * reward

        discount *= GAMMA

        if next_state == "OBSTACLE":
            break

        state = next_state

    return total_return


# ==========================================================
# ACTION VALUE USING TRAJECTORY METHOD
# ==========================================================

def action_value(start_state, first_action):

    # Goal has no action
    if start_state == "G":
        return 0

    state = start_state

    # First action is forced
    next_state = get_next_state(state, first_action)

    reward = get_reward(next_state)

    # First reward
    total_return = reward

    discount = GAMMA

    # If obstacle is reached, episode ends
    if next_state == "OBSTACLE":
        return total_return

    # If goal is reached, episode ends
    if next_state == "G":
        return total_return

    state = next_state

    # Follow the given policy afterwards
    while state != "G":

        action = policy[state]

        next_state = get_next_state(state, action)

        reward = get_reward(next_state)

        total_return += discount * reward

        discount *= GAMMA

        if next_state == "OBSTACLE":
            break

        state = next_state

    return total_return


# ==========================================================
# MAIN PROGRAM
# ==========================================================

print("======================================")
print(" STATE VALUES")
print("======================================")

for state in states:

    if state == "G":
        print("V(G) =", 0)
        continue

    value = state_value(state)

    print("State Value V(" + state + ") =", round(value, 3))


print("\n======================================")
print(" ACTION VALUES")
print("======================================")

for state in states:

    if state == "G":
        continue

    print("\nState:", state)

    for action in actions:

        value = action_value(state, action)

        print(
            "Q(" + state + ", " + action + ") =",
            round(value, 3)
        )
```
