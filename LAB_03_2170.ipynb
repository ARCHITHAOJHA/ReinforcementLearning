Modified ε-Greedy Strategy for Multi-Armed Bandit

```python
import random

# ============================================================
# Part A - Machine Initialization
# ============================================================

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# ============================================================
# Part B - Initialize N, R and Q tables
# ============================================================

# Number of times each machine is played
N = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Total reward obtained from each machine
R = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Estimated Q-value of each machine
Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

# ============================================================
# Suggested Experiment Reward Pattern
# ============================================================

reward_pattern = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Total reward received
total_reward = 0

# Fixed seed for reproducible results
random.seed(42)

# ============================================================
# Part C - Modified Epsilon-Greedy Decision
# ============================================================

for play in range(1, N_play + 1):

    # Generate random number between 0 and 1
    r = random.random()

    # Find machine with highest Q-value
    max_q = max(Q.values())

    highest_machines = [
        machine for machine in machines
        if Q[machine] == max_q
    ]

    # --------------------------------------------------------
    # Exploration
    # --------------------------------------------------------

    if r < epsilon:

        # Exclude the highest Q-value machine
        exploration_machines = [
            machine for machine in machines
            if machine not in highest_machines
        ]

        selected_machine = random.choice(exploration_machines)

        decision = "Exploration"

    # --------------------------------------------------------
    # Exploitation
    # --------------------------------------------------------

    else:

        # Select machine having highest Q-value
        selected_machine = random.choice(highest_machines)

        decision = "Exploitation"

    # ========================================================
    # Part D - Reward Collection
    # ========================================================

    print("\nPlay", play)
    print("--------------------------------")
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Play with machine", selected_machine)

    # Use suggested reward pattern
    index = reward_index[selected_machine]

    if index < len(reward_pattern[selected_machine]):

        reward = reward_pattern[selected_machine][index]

        print("Reward obtained:", reward)

    else:

        # If all predefined rewards are exhausted
        reward = float(
            input("Enter reward: ")
        )

    reward_index[selected_machine] += 1

    # ========================================================
    # Part E - Q-Value Update
    # ========================================================

    # Increase participation count
    N[selected_machine] += 1

    # Add reward to total reward of machine
    R[selected_machine] += reward

    # Incremental sample-average formula
    #
    # Q_new = Q_old + (R - Q_old) / N

    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine])
        / N[selected_machine]
    )

    # Add reward to player's total reward
    total_reward += reward

    print("N(", selected_machine, ") =", N[selected_machine])
    print("R(", selected_machine, ") =", R[selected_machine])
    print(
        "Q(", selected_machine, ") =",
        round(Q[selected_machine], 2)
    )


# ============================================================
# Part F - Display Final Results
# ============================================================

print("\n\n========================================")
print("           FINAL RESULTS")
print("========================================")

print("\nMachine\tN\tR\tQ")
print("----------------------------------------")

for machine in machines:

    print(
        machine,
        "\t",
        N[machine],
        "\t",
        R[machine],
        "\t",
        round(Q[machine], 2)
    )

print("----------------------------------------")

print("Total Reward =", total_reward)

# Find best machine
best_machine = max(Q, key=Q.get)

print("Best Machine =", best_machine)
print("Highest Q-value =", round(Q[best_machine], 2))
```
