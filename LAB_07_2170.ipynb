**Calculate Undiscounted and Discounted Returns**

```python
# Assignment 2: Grid World MDP
# 2 x 3 Deterministic Grid World

class GridWorld:

    def __init__(self):

        # 1. State Space
        self.states = ["S0", "S1", "S3", "S5", "G"]

        # 2. Action Space
        self.actions = ["UP", "DOWN", "LEFT", "RIGHT"]

        # 3. Discount Factor
        self.gamma = 0.9

        # Grid size
        self.rows = 2
        self.cols = 3

        # State positions
        self.positions = {
            "S0": (0, 0),
            "S1": (0, 1),
            "G":  (0, 2),
            "S3": (1, 0),
            "S5": (1, 2)
        }

        # Obstacle position
        self.obstacle = (1, 1)

    # 4. Reward Function
    def get_reward(self, next_state):

        if next_state == "G":
            return 10

        return -1

    # 5. Transition Function
    def get_next_state(self, current_state, action):

        # If already at goal
        if current_state == "G":
            return "G"

        row, col = self.positions[current_state]

        # Calculate new position
        if action == "UP":
            new_row = row - 1
            new_col = col

        elif action == "DOWN":
            new_row = row + 1
            new_col = col

        elif action == "LEFT":
            new_row = row
            new_col = col - 1

        elif action == "RIGHT":
            new_row = row
            new_col = col + 1

        # Invalid action
        else:
            return current_state

        # Check boundaries
        if new_row < 0 or new_row >= self.rows:
            return current_state

        if new_col < 0 or new_col >= self.cols:
            return current_state

        # Check obstacle
        if (new_row, new_col) == self.obstacle:
            return current_state

        # Find corresponding state
        for state, position in self.positions.items():

            if position == (new_row, new_col):
                return state

        return current_state


# ------------------------------------------------
# Main Program
# ------------------------------------------------

# Create environment
env = GridWorld()

# Display Grid
print("================================")
print("       GRID WORLD MDP")
print("================================")

print("\nGrid:")
print("+-----+-----+-----+")
print("| S0  | S1  | G   |")
print("+-----+-----+-----+")
print("| S3  |  X  | S5  |")
print("+-----+-----+-----+")

# Display State Space
print("\nState Space:")
print(env.states)

# Display Action Space
print("\nAction Space:")
print(env.actions)

# Display Discount Factor
print("\nDiscount Factor (Gamma):")
print(env.gamma)

# Take input from user
current_state = input("\nEnter current state: ").strip().upper()

# Validate state
if current_state not in env.states:
    print("Invalid state!")
    exit()

action = input(
    "Enter action (UP/DOWN/LEFT/RIGHT): "
).strip().upper()

# Validate action
if action not in env.actions:
    print("Invalid action!")
    exit()

# Estimate next state
next_state = env.get_next_state(current_state, action)

# Calculate reward
reward = env.get_reward(next_state)

# Display result
print("\n================================")
print("       TRANSITION RESULT")
print("================================")

print("Current State :", current_state)
print("Action        :", action)
print("Next State    :", next_state)
print("Reward        :", reward)

print("================================")
```

**Consider the given 2 × 3 Grid World containing normal states, a
goal state, and an obstacle. Develop a Python program to represent the
environment as an MDP.**

```python
# Assignment 2: 2 x 3 Grid World MDP
# Deterministic Grid World

class GridWorld:

    def __init__(self):

        # a. State Space
        # X is an obstacle, so it is not included
        self.states = ["S0", "S1", "S3", "S4", "G"]

        # b. Action Space
        self.actions = ["UP", "DOWN", "LEFT", "RIGHT"]

        # c. Reward Function
        self.normal_reward = -1
        self.goal_reward = 10
        self.obstacle_reward = -5

        # d. Discount Factor
        self.gamma = 0.9

        # Grid positions
        self.positions = {
            "S0": (0, 0),
            "S1": (0, 1),
            "G":  (0, 2),
            "S3": (1, 1),
            "S4": (1, 2)
        }

        # Obstacle position
        self.obstacle = (1, 0)

    # e. Transition Function
    def transition(self, state, action):

        if state == "S0":

            if action == "UP":
                return "S0"
            elif action == "DOWN":
                return "S0"
            elif action == "LEFT":
                return "S0"
            elif action == "RIGHT":
                return "S1"

        elif state == "S1":

            if action == "UP":
                return "S1"
            elif action == "DOWN":
                return "S3"
            elif action == "LEFT":
                return "S0"
            elif action == "RIGHT":
                return "G"

        elif state == "S3":

            if action == "UP":
                return "S1"
            elif action == "DOWN":
                return "S3"
            elif action == "LEFT":
                return "S3"
            elif action == "RIGHT":
                return "S4"

        elif state == "S4":

            if action == "UP":
                return "G"
            elif action == "DOWN":
                return "S4"
            elif action == "LEFT":
                return "S3"
            elif action == "RIGHT":
                return "S4"

        elif state == "G":
            return "G"

        return state

    # c. Reward Function
    def get_reward(self, state, action):

        # Moving from S0 DOWN attempts to enter obstacle
        if state == "S0" and action == "DOWN":
            return self.obstacle_reward

        # Find next state
        next_state = self.transition(state, action)

        # Reaching goal
        if next_state == "G":
            return self.goal_reward

        # Normal movement
        return self.normal_reward


# =========================================================
# MAIN PROGRAM
# =========================================================

# Create environment
env = GridWorld()

print("==========================================")
print("          2 x 3 GRID WORLD MDP")
print("==========================================")

# Display Grid
print("\nGrid World:")
print("+-------+-------+-------+")
print("|  S0   |  S1   |   G   |")
print("| (0,0) | (0,1) | (0,2) |")
print("+-------+-------+-------+")
print("|   X   |  S3   |  S4   |")
print("| (1,0) | (1,1) | (1,2) |")
print("+-------+-------+-------+")


# a. State Space
print("\nState Space:")
print(env.states)


# b. Action Space
print("\nAction Space:")
print(env.actions)


# c. Reward Function
print("\nReward Function:")
print("Normal Movement :", env.normal_reward)
print("Goal State      :", env.goal_reward)
print("Obstacle        :", env.obstacle_reward)


# d. Discount Factor
print("\nDiscount Factor (Gamma):")
print(env.gamma)


# f. Accept current state
current_state = input(
    "\nEnter current state (S0/S1/S3/S4/G): "
).strip().upper()

if current_state not in env.states:
    print("Invalid state!")
    exit()


# Accept action
action = input(
    "Enter action (UP/DOWN/LEFT/RIGHT): "
).strip().upper()

if action not in env.actions:
    print("Invalid action!")
    exit()


# Estimate next state using Transition Function
next_state = env.transition(current_state, action)


# Calculate reward
reward = env.get_reward(current_state, action)


# Display result
print("\n==========================================")
print("          TRANSITION RESULT")
print("==========================================")

print("Current State :", current_state)
print("Action        :", action)
print("Next State    :", next_state)
print("Reward        :", reward)

print("==========================================")
```
