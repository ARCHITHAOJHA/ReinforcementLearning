Upper Confidence Bound for Multi-Armed Bandit

```python
import math

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Exploration parameter
c = 2

# Reward patterns
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Part B - Initialize knowledge tables
N = [0, 0, 0]       # Number of times each machine is selected
R = [0, 0, 0]       # Total reward from each machine
Q = [0.0, 0.0, 0.0] # Estimated value of each machine
UCB = [float('inf')] * 3

# Keep track of which reward to use next
reward_index = [0, 0, 0]

total_player_reward = 0

print("===== UCB MULTI-ARMED BANDIT ====")

# Part C, D and E
for play in range(1, N_play + 1):

    # Calculate UCB values
    for i in range(3):

        # Select every machine once initially
        if N[i] == 0:
            UCB[i] = float('inf')

        else:
            UCB[i] = Q[i] + c * math.sqrt(math.log(play) / N[i])

    # Choose machine having highest UCB
    selected = UCB.index(max(UCB))

    # Get machine name
    machine = machines[selected]

    # Get reward from given reward pattern
    reward = reward_patterns[machine][reward_index[selected]]

    # Move to next reward for this machine, wrapping around if needed
    reward_index[selected] = (reward_index[selected] + 1) % len(reward_patterns[machine])

    print(f"\nPlay {play}")
    print(f"Selected Machine: {machine}")
    print(f"Reward: {reward}")

    # Update participation count
    N[selected] += 1

    # Update total reward of selected machine
    R[selected] += reward

    # Incremental sample-average update
    Q[selected] = Q[selected] + (1 / N[selected]) * (reward - Q[selected])

    # Update total reward received by player
    total_player_reward += reward

    print(f"N = {N}")
    print(f"R = {R}")
    print(f"Q = {[round(x, 2) for x in Q]}")

# Recalculate final UCB values
for i in range(3):

    if N[i] == 0:
        UCB[i] = float('inf')

    else:
        UCB[i] = Q[i] + c * math.sqrt(math.log(N_play) / N[i])


# Part F - Display final results
print("\n\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\t\tQ")

for i in range(3):
    print(
        f"{machines[i]}\t"
        f"{N[i]}\t"
        f"{R[i]}\t"
        f"{UCB[i]:.2f}\t\t"
        f"{Q[i]:.2f}"
    )

print("\nTotal Reward received by player =", total_player_reward)

print(
    "Average Reward per play =",
    round(total_player_reward / N_play, 2)
)
```
